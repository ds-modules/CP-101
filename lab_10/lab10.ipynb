## Data cleaning

```python
alameda = pd.read_csv("alameda_sample.csv")
alameda.head()
```

```python
len(alameda)
```

```python
# Clean the dataset of missing values
# Display rows with missing values in any column. There are a total of 81.
alameda[alameda.isnull().any(axis=1)].head()
```

```python
# Get rid of the rows where home_tract is NaN
# This code actually drops any row where there is a missing value in any column, not just home_tract 
alameda = alameda.dropna(axis=0, how='any')
len(alameda)
```

```python
# Now if you check for rows with missing values:
alameda[alameda.isnull().any(axis=1)]
```
