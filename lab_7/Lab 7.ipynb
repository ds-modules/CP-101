In Python we can represent strings with either double quotes " " or single quotes ' ' — it doesn't matter which. But in case you need to use a single quote inside your string (e.g. Sam's), use double quotes on the outside or use backslash before the single quote (e.g. 'Sam\'s').

### Lists

A list is a sequence of objects. Lists are the datatypes that allow us to unite many different datatypes together which makes them very useful. 

To create an empty list simply type in:

```python
my_list = []
```

The objects in a list have a set order. The counting in a list starts with 0 and ends with n-1, where n is the total number of elements in a list. Now let's create a list of different elements and try to play with it.

```python
new_lst = [1, 15, 29, "Berkeley", 6, [4,5]]

#As we mentioned before, lists can have different elements all together. They can even contain other lists.
```

To access an element at a specific index, we will use:

```python
new_lst[1]

#Remember that counting starts with 0, which means that the 0th element in our list is 1 and the 1st element is 15.
```

You can also replace certain elements in our list. To do that we will use the same notation as shown above to access an element at a specific index and then we will make it equal something. Like so:

```python
new_lst[1] = 20
new_lst
```

We can also copy chunks of our intial list into a new list. 

```python
my_lst = new_lst[1:4]
my_lst

#Remember that the last element index is exclusive, so always put n+1 at the end, 
#where n is your desired last index from the list
```

We can also access our list from a specific index onwards or up to a specific index. Run the cells below to see it for yourself.

```python
new_lst[2:]
```

```python
new_lst[:3]
```

Another operation we can do with ":" is to call a list skipping every n-th element.

```python
new_lst[::2]

#As you can see, now it only returns every 2nd element from our list.
```

There are two popular methods that can remove elements from a list. they are pop() and remove(). pop() removes the last element from a list and returns it in that cell. remove() can take out any element from a list, but you need to specify the element itself, not an index.

```python
#Let's make a list of Berkeley restaurants for this exercise. 
#Feel free to change the restaurants.

berkeley = ["Gypsy's", 'Homemade Cafe', '1951 Coffee', "Bear Ramen", "Flying Falafel"]
berkeley
```

Now let's use pop() to save the last element of our list into a new variable.

```python
last_elem = berkeley.pop()
last_elem
```

```python
berkeley

#As you can see, our list "berkeley" doesn't have "Flying Falafel" anymore.
```

Now let's use remove() to remove a specific restaurant from the list.

```python
berkeley.remove("Bear Ramen")
berkeley
```

The other useful elements that can be used with Python lists are append(), extend(), and insert(). Let's see how they can be used in the cells below:

```python
berkeley.append("Cheese Board")
berkeley

#append() just adds an element to your list.
```

```python
berkeley.insert(0, "Platano Cafe")
berkeley

#For the insert() method you need to specify the location in the list first, and then specify an element.
```

```python
telegraph = ["Chipotle", "Sliver"]
berkeley.extend(telegraph)
berkeley

# extend() only accepts other lists as its argument. 
```
